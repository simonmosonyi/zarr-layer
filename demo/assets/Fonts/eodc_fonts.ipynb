Setting the font for all Markdown text fields

```python
from IPython.core.display import HTML

HTML("""
<style>
@font-face {
  font-family: 'Mark OT';
  src: url('/home/koenifra/fonts/MarkOT/MarkOT.otf') format('opentype');
}
body {
  font-family: 'Mark OT', sans-serif;
}
</style>
""")
```

Prepare the font for the plots

```python
title_size = 20
label_size = 14
ticks_size = 12
```

```python
font_path = "/home/koenifra/fonts/MarkOT/MarkOT.otf"
font_path_italic = "/home/koenifra/fonts/MarkOT/MarkOT-Italic.otf"
font_path_bold = "/home/koenifra/fonts/MarkOT/MarkOT-Bold.otf"
```

```python

eodc_title_regular = fm.FontProperties(fname=font_path, size=title_size)
eodc_title_italic = fm.FontProperties(fname=font_path_italic, size=title_size)
eodc_title_bold = fm.FontProperties(fname=font_path_bold, size=title_size)

eodc_label = fm.FontProperties(fname=font_path, size=label_size)
eodc_ticks = fm.FontProperties(fname=font_path, size=ticks_size)
```

```python
plt.figure(figsize=(8, 4))
plt.text(0.1, 0.8, 'Regular Font', fontproperties=eodc_title_regular)
plt.text(0.1, 0.6, 'Bold Font', fontproperties=eodc_title_bold)
plt.text(0.1, 0.4, 'Italic Font', fontproperties=eodc_title_italic)
plt.text(0.1, 0.2, 'Label Font', fontproperties=eodc_label)
plt.text(0.1, 0.0, 'Tick Font', fontproperties=eodc_ticks)

plt.axis('off')
plt.title('MarkOT Font Styles')
plt.show()
```
